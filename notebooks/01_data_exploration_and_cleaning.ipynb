# 01. Data Exploration & Cleaning
**Deskripsi:** Tahap awal pemeriksaan dataset Tokopedia Product Reviews, pembersihan data, dan handling missing values.


```python
# 1. Import Library (pandas, numpy, dll)
import os 
import pandas as pd 
import numpy as np
```

```python
# 2. Load Raw Dataset dari 'data/raw/tokopedia_products_raw.csv'
raw_path = "../data/raw/tokopedia_products_raw.csv"
df = pd.read_csv(raw_path)
```

```python
# 3. Check Dataset Info (shape, dtypes, head, tail)
print("---Ringkasan Dataset---")
print(f"Jumlah Baris dan Kolom: {df.shape}")
print("/n---5 Baris Pertama Dataset---")
print(df.head())
print("/n---Informasi Tipe Data")
print(df.info())
print("/n--- Jumlah Missing Value")
print(df.isnull().sum())
```

```python
# 4. Handling Missing Values & Duplicates
df_clean = df.drop_duplicates().copy()
df_clean = df_clean.dropna(subset=['rating','category','product_name'])
df_clean['rating'] = pd.to_numeric(df_clean['rating'], errors='coerce')
df_clean['sold'] = pd.to_numeric(df_clean['sold'], errors='coerce').fillna(0).astype(int)

def classify_sentiment(rating):
    if rating >=4:
        return 'Positive'
    elif rating ==3:
        return 'Neutral'
    else:
        return 'Negative'

df_clean['sentinent_category'] = df_clean['rating'].apply(classify_sentiment)
```

```python
# 5. Data Type Formatting (rating -> float, sold -> int)
output_path = "../data/processed/tokopedia_cleaned.csv"
os.makedirs("../data/processed", exist_ok=True)
df_clean.to_csv(output_path, index=False)
print(f"Data berhasil dibersihkan! total baris akhir: {len(df_clean)}")
```

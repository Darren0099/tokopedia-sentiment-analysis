# 04. Executive Summary & Dashboard Preparation
**Deskripsi:** Rangkuman insight utama untuk kebutuhan slide presentasi dan dashboard Looker Studio / Power BI.


```python
# 1. Import All Processed Insights & Metrics Summary
import pandas as pd

df = pd.read_csv("../data/processed/tokopedia_cleaned.csv")
```

```python
# 2. Rekap KPI Utama (Total Sold, Avg Rating, Total Categories, Total Shops)
total_sold = df['sold'].sum()
avg_rating = df['rating'].mean()
total_products = df['product_id'].nunique()
total_shops = df['shop_id'].nunique()

kpi_df = pd.DataFrame([{
    'Total_Items_Sold': total_sold,
    'Average_Rating': round(avg_rating, 2),
    'Total_Products': total_products,
    'Total_Shops': total_shops
}])

print("--- Key Performance Indicators (KPIs) ---")
print(kpi_df)
```

```python
# 3. Buat Ringkasan Tabel Agregasi untuk Dashboard Looker Studio / Power BI
# normalize the sentiment column name because the source data uses "sentinent_category"
if 'sentiment_category' not in df.columns and 'sentinent_category' in df.columns:
    df = df.rename(columns={'sentinent_category': 'sentiment_category'})

summary_by_cat = (
    df.groupby(['category', 'sentiment_category'], as_index=False)
      .agg(
          total_sold=('sold', 'sum'),
          avg_rating=('rating', 'mean'),
          product_count=('product_id', 'count')
      )
)

summary_by_cat.to_csv("../outputs/reports/summary_metrics.csv", index=False)
kpi_df.to_csv("../outputs/reports/kpi_metrics.csv", index=False)

print("\nSeluruh pengolahan selesai! File grafik tersimpan di 'outputs/figures/' dan laporan CSV tersimpan di 'outputs/reports/'.")
```
